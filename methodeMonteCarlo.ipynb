## MethodeMonteCarlo

### 1. choix des coordonnées x et y entre -1 et 1  en utilisant random.uniform

```python
Ox = 0
Oy = 0
R = 1
```

```python
x = random.uniform(-1, 1) 
y = random.uniform(-1, 1)
```

### 2. Calcul de la distance du centre du cercle et du point de Coordonnées x et y

```python
d = math.sqrt((x-Ox)**2+(y-Oy)**2)
```

```python
print("La distance entre le centre  du cercle et le point est : ", d)
```

### 3. verifions si le point appartient au cercle

```python
if d <= R:
    print("le point appartient au cercle ")
else :
    print("le point n'appartient pas au cercle ")
```

### 4. Incrémentation du compteur n

```python
n = 0
N = 100 ### on peut aussi faire pour 1000 ou 10000
for i in range(N):
    x = random.uniform(-1, 1) 
    y = random.uniform(-1, 1)
    d = math.sqrt((x-Ox)**2+(y-Oy)**2)
    if d<= R:
        n = n+1
        
```
